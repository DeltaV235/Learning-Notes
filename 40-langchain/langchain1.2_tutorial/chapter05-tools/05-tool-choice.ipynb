```python
from dotenv import load_dotenv
from langchain.chat_models.base import init_chat_model
from langchain_core.messages import HumanMessage
from langchain_core.tools import tool
from rich import print as rprint
```

```python
load_dotenv(override=True)

model = init_chat_model(model="deepseek:deepseek-v4-flash")


@tool(parse_docstring=True)
def get_weather(city: str) -> str:
    """
    获取当日天气

    Args:
        city: 城市名称
    """
    return f'{city}当天晴朗'


model_with_tools = model.bind_tools(tools=[get_weather], tool_choice="none")

response = model_with_tools.invoke([HumanMessage(content="上海今天天气如何？")])

rprint(response)
```

```python
model_with_tools = model.bind_tools(tools=[get_weather], tool_choice="auto")

response = model_with_tools.invoke([HumanMessage(content="上海今天天气如何？")])
rprint(response)
```

```python
model = init_chat_model(model="openrouter:openai/gpt-5.6-luna")
model_with_tools = model.bind_tools(tools=[get_weather], tool_choice="required")

response = model_with_tools.invoke([HumanMessage(content="上海今天天气如何？")])
rprint(response)
```

```python
model = init_chat_model(model="openrouter:openai/gpt-5.6-luna")
model_with_tools = model.bind_tools(tools=[get_weather], tool_choice="required")

response = model_with_tools.invoke([HumanMessage(content="1 + 2 = ？")])
rprint(response)
```

```python
@tool(parse_docstring=True)
def get_weather1(city: str) -> str:
    """
    获取当日天气

    Args:
        city: 城市名称
    """
    return f'{city}当天晴朗'


@tool(parse_docstring=True)
def get_weather2(city: str) -> str:
    """
    获取当日天气

    Args:
        city: 城市名称
    """
    return f'{city}当天晴朗'


model_with_tools = model.bind_tools([get_weather1, get_weather2], tool_choice="get_weather2")

messages = [
    HumanMessage("杭州今天天气如何？")
]

response = model_with_tools.invoke(messages)
rprint(response)
```

```python
model_with_tools = model.bind_tools([get_weather1, get_weather2], tool_choice="get_weather1")

messages = [
    HumanMessage("杭州今天天气如何？")
]

response = model_with_tools.invoke(messages)
rprint(response)
```
