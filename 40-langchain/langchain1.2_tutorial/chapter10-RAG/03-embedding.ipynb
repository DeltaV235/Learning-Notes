# Embed

```python
import os

from dotenv import load_dotenv
from langchain.embeddings import init_embeddings
from langchain_community.document_loaders import CSVLoader
```

```python
load_dotenv(override=True)
embedding_model_qwen = init_embeddings(
    model="openai:qwen/qwen3-embedding-4b",
    base_url=os.getenv("OPENROUTER_API_BASE"),
    api_key=os.getenv("OPENROUTER_API_KEY"),
    check_embedding_ctx_length=False,
    dimensions=2048,
)

embedding_model_nvidia = init_embeddings(
    model="openai:nvidia/nemotron-3-embed-1b:free",
    base_url=os.getenv("OPENROUTER_API_BASE"),
    api_key=os.getenv("OPENROUTER_API_KEY"),
    check_embedding_ctx_length=False,
    model_kwargs={"encoding_format": "float"},
)

embedding_model_openai = init_embeddings(
    model="openai:openai/text-embedding-3-large",
    base_url=os.getenv("OPENROUTER_API_BASE"),
    api_key=os.getenv("OPENROUTER_API_KEY"),
)
```

```python
text = "What was the name mentioned in the conversation?"

result = embedding_model_qwen.embed_query(text)

print(len(result))
print(result[:10])
```

```python
result = embedding_model_openai.embed_query(text)

print(len(result))
print(result[:10])
```

```python
result = embedding_model_nvidia.embed_query(text)

print(len(result))
print(result[:10])
```

```python
loader = CSVLoader("./asset/load/02-load.csv", encoding="utf-8")
docs = loader.load_and_split()

document = [text.page_content for text in docs]

result = embedding_model_openai.embed_documents(document)

for index, embed_result in enumerate(result):
    print("-" * 20)
    print(len(embed_result))
    print(f"{document[index]}:\n {embed_result[:10]}")
```
